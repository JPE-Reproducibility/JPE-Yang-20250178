western

```python
import os
import pandas as pd

# directory path
dir_path = 'PATH_TO_ENTTEMPLATES_DATA_ROOT/Analysis/python_Similarity/processed_western'

# create an empty list to collect dataframes
df_list = []
counter = 0

for filename in os.listdir(dir_path):
    if filename.endswith('.csv'):
        counter += 1
        # parse fullname from the filename
        fullname = filename.replace('similarity_ST_', '').replace('_processed_western.csv', '')

        # create the complete filepath
        file_path = os.path.join(dir_path, filename)
        
        # read the csv file and add it to the list
        df = pd.read_csv(file_path)
        
        # add a new column 'fullname'
        df['fullname'] = fullname

        df_list.append(df)
        print(f'finished processing {filename}'+str(counter))
# concatenate all the dataframes in the list
full_df = pd.concat(df_list, ignore_index=True)
full_df.to_csv('PATH_TO_ENTTEMPLATES_DATA_ROOT/Analysis/python_Similarity/similarity_compiled_western.csv')

```
